# Part of Speech (POS) Tagging Example
This notebook contains a function to perform POS tagging on English sentences using the `nltk` library.

```python
# Install nltk if not already installed
# !pip install nltk

import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag

# Download necessary NLTK data (runs quietly)
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('averaged_perceptron_tagger', quiet=True)
nltk.download('averaged_perceptron_tagger_eng', quiet=True)

def get_pos_tags(text):
    """
    Takes an English sentence and returns a list of tuples containing (word, POS tag).
    """
    # Tokenize the sentence into words
    tokens = word_tokenize(text)
    
    # Perform Part-of-Speech tagging
    tags = pos_tag(tokens)
    
    return tags

# Example usage
sample_text = "The quick brown fox jumps over the lazy dog."
result = get_pos_tags(sample_text)

print(f"Input Text: {sample_text}\n")
print("POS Tags (Word : Tag):")
print("-" * 25)
for word, tag in result:
    print(f"{word:12} : {tag}")
```
